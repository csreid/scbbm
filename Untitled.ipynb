```python
%load_ext autoreload
%autoreload 2
#from main_loop import _collate
from fg_dataset import PlayByPlayDataset
from torch.utils.data import DataLoader
from torch.nn.utils.rnn import pad_sequence
import torch
from models.game_sim import GameSimulator
from torch.distributions.categorical import Categorical

def _collate(batch):
    teams = [r[0] for r in batch]
    plays = [r[1] for r in batch]

    teams_tensor = torch.stack(teams, dim=0)

    plays_tensor = pad_sequence(plays)
    return teams_tensor.unsqueeze(0).expand(plays_tensor.shape[0], -1, -1), plays_tensor


ds = PlayByPlayDataset('cleaned_dataset.csv', subgame_len=20)
loader = DataLoader(ds, batch_size=8, collate_fn=_collate)
model = GameSimulator(len(ds.teams), len(ds.plays))
model = model.to('cuda:0')
model.load_state_dict(torch.load('model.ptch'))
```

```python
def sim_game(model, ds, home_team, away_team):
    teams_vec = ds._teams_to_tensor(home_team, away_team)
    first_play = 'Jump Ball won by Rutgers'
    simmed_plays = [ds.play_ids_map[first_play]]#, ds.play_ids_map['Zach Edey made Layup.']]
    h = None
    for _ in range(100):
        teams, plays = _collate([[teams_vec, torch.tensor(simmed_plays)]])
        out, _ = model(teams, plays, None, state=h)
        next_play_idx = Categorical(torch.nn.functional.softmax(out[-1, 0], dim=0)).sample().item()
        #next_play_idx = torch.argmax(out[-1, 0], dim=0)
        #print(torch.nn.functional.softmax(out[-1,0], dim=0)[next_play_idx])
        next_play = ds.plays[next_play_idx]
        simmed_plays.append(next_play_idx)

    return [ds.plays[s] for s in simmed_plays]

def sim_next(model, ds, home_team, away_team, plays):
    teams_vec = ds._teams_to_tensor(home_team, away_team)
    teams_vec, plays_vec = _collate([[teams_vec, torch.tensor(plays)]])

    out, _ = model(teams_vec, plays_vec, None)

    return out
```

```python
import numpy as np
import matplotlib.pyplot as plt

first_play = 'Jump Ball won by Rutgers'
simmed_plays = [ds.play_ids_map[first_play]]#, ds.play_ids_map['Zach Edey made Layup.']]

out = sim_next(model, ds, 'Purdue', 'Rutgers', simmed_plays)[-1][0]
plt.scatter(np.arange(len(out)), torch.nn.functional.softmax(out, dim=0).cpu().detach().numpy(), np.ones(len(out))/10, alpha=0.5,)
plt.show()
```
